# "@" Decorator

```python
# a decorator is a function that takes another function as an argument and extends the behavior of the latter function without explicitly modifying it. Decorators are often used in Python to add functionality to existing functions or methods in a clean and readable way. They are denoted by the @ symbol followed by the name of the decorator function, placed above the function definition that they are meant to modify. Decorators can be used for various purposes, such as logging, timing, access control, and more.
```

```python
class Animal:
    @property
    def show(self):
        print("Hello how are you?")

obj = Animal()
obj.show
```

```python
def decorate(func):
    def wrapper():
        print("I will print myself before the function is called.")
        func()
        print("I will print myself after the function is called.")
    return wrapper

@decorate
def hello():
    print("Hello, world!")

hello()

# the @decorate syntax is a shorthand for applying the decorate function to the hello function. It is equivalent to writing hello = decorate(hello). When we call hello(), it will execute the wrapper function defined inside the decorate function, which will print the messages before and after calling the original hello function.
```

```python
def decorate(func):
    def wrapper(a,b):
        print("The addition to your function are")
        func(a,b)
        print("I hope you liked it")
    return wrapper

@decorate
def hello(a,b):
    print(f"The total is {a + b}")

hello(12,67)
```

```python
# for making decorators with arguments it is difficult to write, so we use *args and **kwargs to make it more flexible and easier to use. *args allows us to pass a variable number of non-keyword arguments to the wrapper function, while **kwargs allows us to pass a variable number of keyword arguments. This way, we can create decorators that can work with any function regardless of the number of arguments it takes.
```

```python
def addition(a,b):
    print(f"The total is {a + b}")

addition(12,67) 
```

# *Args (Arguments)

```python
# addition(12,67,72) 
# this will fail because the addition function is defined to take only two arguments, but we are trying to pass three arguments. To make it work with any number of arguments, we can modify the addition function to use *args:

# All the arguments passed to the function will be collected into a tuple called args, and we can use the sum() function to calculate the total of all the arguments.

def addition(*args):
    total = sum(args)
    print(f"The total is {total}")

addition(1,213,12,31,31,231)
```

# ** Kwargs (Keyword Arguments)

```python
# Kwargs allows us to pass a variable number of keyword arguments to a function. The keyword arguments are passed as a dictionary, where the keys are the argument names and the values are the argument values. This can be useful when we want to create functions that can accept a flexible number of named parameters.

def information(**kwargs):
    print("Your info is : \n")

    for i in kwargs:
        print(f"{i} : {kwargs[i]}")


information(name="Dhruval", age=20, city="Surat")
```

# use case of Args and Kwargs in Decorators

```python
def decorate(func):
    def wrapper(*args, **kwargs):
        print("The addition to your function are")
        func(*args, **kwargs)
        print("I hope you liked it")
    return wrapper

@decorate
def hello(a,b):
    print(f"The total is {a + b}")
hello(12,67)

```

# Ternary Operations

```python
# Ternary operator is a one line if else statement. It is used to assign a value to a variable based on a condition. The syntax is: variable = value_if_true if condition else value_if_false

a = 12 

print("Even") if a % 2 == 0 else print("Odd")
```

# Comprehensions

```python
# list comprehension is a concise way to create lists in Python. It allows you to generate a new list by applying an expression to each item in an iterable, while optionally filtering items using a condition. The syntax for list comprehension is: [expression for item in iterable if condition]. This can be more efficient and readable than using traditional loops to create lists.
l = [i for i in range(1,21) if i % 2 == 0]
print(l)
```

```python
# Dictionary comprehension is similar to list comprehension, but it creates a dictionary instead of a list. The syntax for dictionary comprehension is: {key_expression: value_expression for item in iterable if condition}. This allows you to create dictionaries in a concise and readable way.

l = {i: i**2 for i in range(1,11) if i % 2 == 0}
print(l)
```

```python
# set comprehension is similar to list comprehension, but it creates a set instead of a list. The syntax for set comprehension is: {expression for item in iterable if condition}. This allows you to create sets in a concise and readable way.

s = {i for i in range(1,21) if i % 2 == 0}
print(s)
```

# Lambda Functions

```python
# a lambda function is a small anonymous function that can take any number of arguments, but can only have one expression. The syntax for a lambda function is: lambda arguments: expression. Lambda functions are often used for short, simple functions that are not worth defining with a full function definition. They are commonly used in conjunction with higher-order functions like map(), filter(), and reduce().

addition = lambda a,b: a+b

print(addition(12,67))

addition = lambda a: "even" if a % 2 == 0 else "odd"

print(addition(12))
```

# Map

```python
# map is used for applying a function to all the items in an iterable (like a list) and returning a new iterable with the results. The syntax for map is: map(function, iterable). This can be useful for transforming data in a concise and efficient way.

numbers = [1,2,3,4,5]

result = lambda x: x*2
squared_numbers = list(map(result, numbers))

print(squared_numbers)
```

```python
a = [1,2,3,4,5]

def double(x):
    return x*2

result = list(map(double, a))

print(result)
```

# Filter

```python
# filter is used to filter items in an iterable based on a function that returns a boolean value. The syntax for filter is: filter(function, iterable). This can be useful for selecting specific items from a collection based on certain criteria.

numbers = [1,2,3,4,5]

def even(x):
    if x % 2 == 0:
        return True
    else:
        return False
    

even_numbers = list(filter(even, numbers))
print(even_numbers)
```

```python
numbers = [1,2,3,4,5]

result = list(filter(lambda x: x % 2 == 0, numbers))
print(result)
```
