# Attributes

```python
# types of attributes

# variables defined inside the class and outside any method are called class variables. They are shared among all instances of the class.

# variables defined inside the __init__ method and prefixed with self are called instance variables. They are unique to each instance of the class and are used to store data that is specific to that instance.

class Animal:
    name = "Lion"

    def __init__(self, age):
        self.age = age # instance attribute / variable


    def show (self): # instance method
        print("How are you doing?")

    # @ decorator is used to modify the behavior of a function or method. It is a way to wrap another function or method in order to extend its behavior without modifying the original function or method.

    @classmethod

    # class method is a method that is bound to the class and not the instance of the class. It can access and modify class state that applies across all instances of the class. It takes cls as the first parameter which refers to the class itself.

    def hello(cls):
        print("Hello how are you? I am a class method")


    @staticmethod
    def static_hello():
        print("Hello how are you? I am a static method")


obj = Animal(12)

obj.show() # This will call the instance method show using the instance obj. The instance method can be called on an instance, and it will receive the instance as the first argument (self).

obj.hello() # This will call the class method hello using the instance obj. The class method can be called on an instance, and it will receive the class as the first argument (cls) instead of the instance (self).

obj.static_hello() # This will call the static method static_hello using the instance obj. Static methods can be called on an instance, but they do not receive any special first argument (neither self nor cls) and behave like regular functions that belong to the class's namespace.


obj = Animal(12)

print(obj.name) # This will access the class variable name using the instance obj. Since name is a class variable, it is shared among all instances of the class, and we can access it through any instance or directly from the class itself.

print(obj.age) # This will access the instance variable age using the instance obj. Since age is an instance variable, it is unique to each instance of the class, and we can only access it through an instance of the class.
```

# Inheritance

```python
# Inheritance is a fundamental concept in object-oriented programming that allows a new class (called a child or subclass) to inherit attributes and methods from an existing class (called a parent or superclass). This promotes code reusability and establishes a natural hierarchical relationship between classes.


class Factorymumbai: # This is a class that represents a factory in Mumbai. It has an attribute a and a method hello.
    
    a = "I am an attribute mentioned inside a factory"

    def hello(self):
        print("How are you?")

class Factorypune(Factorymumbai): # This is a child class that inherits from the Factorymumbai class. It will have access to all the attributes and methods of the Factorymumbai class.

    def hello(self): # This is an overridden method. It will override the hello method of the parent class Factory.
        print("I am a factory in Pune")


obj = Factorymumbai() # creating an instance of the Factorymumbai class and assigning it to the variable obj

print(obj.a)


obj2 = Factorypune() # creating an instance of the Factorypune class and assigning it to the variable obj2


print(obj2.hello())
```

# Constructor

```python
class Animal:
    def __init__(self,name):
        self.name = name

    def show(self):
        print(f"Hello your name is {self.name}")


class Human(Animal):

    def __init__(self, name, age):
        super().__init__(name)
         # This is calling the constructor of the parent class (Animal) to initialize the name attribute. The super() function is used to call the parent class's method, and we pass the name argument to it. This allows us to reuse the initialization logic of the parent class while also adding additional attributes or functionality in the child class (Human).
        self.age = age

    def show(self):
        print(f"Hello your name is {self.name} and you are {self.age} years old.")



person1 = Human("Dhruval", 20)
 # creating an instance of the Human class and passing the name "Dhruval" and age 20 to the constructor of the Animal class through inheritance

Animal1 = Animal("Tiger")
 # creating an instance of the Animal class and passing the name "Tiger" to the constructor

Animal1.show()

person1.show() 
# This will call the show method of the Animal class using the instance person1 of the Human class. Since Human inherits from Animal, it has access to the show method, and it will print "Hello your name is Dhruval".

```

# multilevel inheritance

```python
class Factory():
    def __init__(self, material, zips):
        self.material = material
        self.zips = zips

class BhopalFactory(Factory):

    def __init__(self, material, zips, colors):
        super().__init__(material, zips)
         # This is calling the constructor of the parent class (Factory) to initialize the material and zips attributes. The super() function is used to call the parent class's method, and we pass the material and zips arguments to it. This allows us to reuse the initialization logic of the parent class while also adding additional attributes or functionality in the child class (BhopalFactory).
        self.colors = colors

class PuneFactory(BhopalFactory):

    def __init__(self, material, zips, colors, pockets):
        super().__init__(material, zips, colors)
         # This is calling the constructor of the parent class (BhopalFactory) to initialize the material and zips attributes. The super() function is used to call the parent class's method, and we pass the material and zips arguments to it. This allows us to reuse the initialization logic of the parent class while also adding additional attributes or functionality in the child class (PuneFactory).
        self.pockets = pockets
```

# Polymorphism

```python
def show():
    print("How are you")

def show():
    print("You are best")

show()

# here the function is being overridden. The second definition of the show function will overwrite the first one, so when we call show(), it will execute the second definition and print "You are best".
```

# method overriding

```python
class Animal:
    def show(self):
        print("How are you doing?")

class Human(Animal):
    def show(self):
        print("I am a human")


obj = Human()
obj.show()

# this is method overriding. The show method in the Human class overrides the show method in the Animal class. When we create an instance of the Human class and call the show method, it will execute the show method defined in the Human class, which prints "I am a human".
```

### method overloading does not exist in python.

```python
# Duck Typing


class Animal:
    def show(self):
        print("How are you doing?")

class Human():
    def show(self):
        print("I am a human")

obj1 = Animal()
obj2 = Human()

obj1.show() # This will call the show method of the Animal class using the instance obj1. It will print "How are you doing?".

obj2.show() # This will call the show method of the Human class using the instance obj2. It will print "I am a human".
```

# Abstractions

```python
# abstraction does not exist in python

# though the concept is there but it is not enforced by the language. We can achieve abstraction in python using abstract base classes (ABCs) from the abc module, but it is not a built-in feature of the language like in some other programming languages.
```

```python
# Abstract classes are the classes that cannot be instantiated and are meant to be subclassed. They can contain abstract methods, which are methods that are declared but not implemented in the abstract class. Subclasses of the abstract class are expected to provide implementations for the abstract methods.
```

```python
class Square:
    def __init__(self, side_length):
        self.side_length = side_length

class Circle:
    def __init__(self, radius):
        self.radius = radius
```

```python
from abc import ABC, abstractmethod

class abstract (ABC):
    @abstractmethod
    def show(self):
        pass

    def perimeter(self):
        pass

    def area(self):
        pass

class Square(abstract):
    def __init__(self, side):
        self.side = side

    def show(self):
        print("This is a square")

    def perimeter(self):
        return 4 * self.side

    def area(self):
        return self.side ** 2

class Circle(abstract):
    def __init__(self, radius):
        self.radius = radius

    def show(self):
        print("This is a circle")

    def perimeter(self):
        return 2 * 3.14 * self.radius

    def area(self):
        return 3.14 * self.radius ** 2

obj = Circle(5)
obj.show()
print("Perimeter:", obj.perimeter())
print("Area:", obj.area())
```

# Dunder Methods

```python
# Dunder methods are the methods which starts with __ and ends with __. They are also called magic methods or special methods. They are used to define the behavior of an object when it is used in a certain way. For example, the __str__ method is used to define the string representation of an object, the __add__ method is used to define the behavior of the + operator when applied to objects of a class, and so on. Dunder methods allow us to customize the behavior of our classes and make them more intuitive to use.
```

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Animal name is {self.name}"


obj = Animal("Tiger")

print(obj) # This will call the __str__ method of the Animal class and print "Animal name is Tiger". The __str__ method is a dunder method that defines the string representation of the object when it is printed or converted to a string.
```

```python
# __add__


class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __add__(self, other):
        return f"{self.name} and {other.name} are friends and ages are {self.age} and {other.age} respectively"
    
    #self refers to the instance of the class that is calling the method, while other refers to the instance of the class that is being added to self. The __add__ method takes two instances of the Animal class and returns a string that describes their relationship.

    #other refers to the second operand in the addition operation. When we use the + operator between two instances of the Animal class, the __add__ method is called, and other will refer to the second instance that is being added to the first instance (self).

obj1 = Animal("Tiger", 5)
obj2 = Animal("Lion", 3)
obj3 = Animal("Elephant", 34) # we cannot add three objects together because the __add__ method is defined to take only two parameters (self and other). If we try to add three objects together, it will result in a TypeError because the __add__ method does not know how to handle more than two operands.

print(obj1 + obj2) # This will call the __add__ method of the Animal class and print "Tiger and Lion are friends". The __add__ method is a dunder method that defines the behavior of the + operator when used with objects of the class.
```
