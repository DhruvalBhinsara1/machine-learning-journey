# **Advanced Array Manipulation**

### Stacking Arrays

```python
import numpy as np
```

```python
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
```

```python
np.vstack((a,b))
```

```python
np.hstack((a,b))
```

```python
np.column_stack((a,b))
```

## Splitting Arrays

```python
c = np.arange(16).reshape(4,4)
```

```python
c
```

```python
np.hsplit(c,2)
```

```python
# vsplit and split with axis = 0 are the same.

np.split(c,2)
```
