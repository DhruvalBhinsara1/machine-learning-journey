# Arithmetic operations

*Note*: size of the array must be same for both the arrays

```python
a1 = np.array([1,2,3])
a2 = np.array([4,5,6])
```

```python
a1 + a2
```

```python
a1 - a2
```

```python
a1 * a2
```

```python
a1 / a2
```

```python
a1 // a2
```

```python
a1 ** a2
```
