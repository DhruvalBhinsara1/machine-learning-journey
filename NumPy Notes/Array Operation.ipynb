## Array Broadcasting

```python
# Array broadcasting works on the principle of the smaller array being "broadcast" across the larger array so that they have compatible shapes. This allows for element-wise operations to be performed on arrays of different sizes and shapes without the need for explicit replication of data.

# we can add a scalar value to an array, and the scalar will be broadcasted to each element of the array.
```

```python
l = [10,20,30,40]

arr = np.array(l)
```

```python
arr + 10
```

```python
arr2 = np.arange(1,26).reshape(5,5)
arr2
```

```python
arr2 + 10
```

```python
arr2 * 2
```
