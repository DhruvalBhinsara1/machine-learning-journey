## **Difference Between a list and array in Python**

### 1. Type of Elements:
 - List: A list can contain elements of different data types (e.g., integers,
   strings, floats, etc.).
 - Array: An array typically contains elements of the same data type (e.g.,
   all integers or all floats).

### 2. Performance:
 - List: Lists are more flexible but can be slower for large datasets due to
   their dynamic nature.
 - Array: Arrays are more efficient for numerical operations and can be faster
   for large datasets, especially when using libraries like NumPy.

### 3. Functionality:
 - List: Lists have a wide range of built-in methods for manipulation (e.g.,
   append, remove, etc.).
 - Array: Arrays have fewer built-in methods but can be manipulated using
   libraries like NumPy, which provide extensive functionality for array operations.

---

## **Array Methods**

````python
arr
````

````python
# Minimum
print(arr.min())
````

````python
# Maximum
print(arr.max())
````

````python
# this is the sum using the inbuilt function of python
print(arr.sum())
````

---
### **Note**:

#### `Row is the` **`1`** `Axis`
#### `Column is the` **`0`** `Axis`
---

````python
# this is the sum using axis argument

print(arr.sum(axis=0)) # column wise sum
print(arr.sum(axis=1)) # row wise sum
````

````python
# Mean/Average
print(np.mean(arr))
````

````python
# Standard Deviation
print(np.std(arr))
````

````python
# Argmin and Argmax
print(arr.argmin())
print(arr.argmax())

# this is used to retrieve the poisition of the minimum and maximum value in the array. It returns the index of the minimum and maximum value in the array.
````
