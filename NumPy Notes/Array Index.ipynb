# **Importing NumPy**

```python
import numpy as np
```

## **numpy indexing and slicing and slicing of vectors**

```python
arr = np.arange(11,21)
arr
```

```python
# indexing

arr[9]
```

```python
# slicing

# format (start:stop:step)

arr[0:5]
```

## **numpy indexing and slicing and slicing of matrix**

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0]
```

```python
#finding 1

arr[0,0]
```

```python
#finding 5

arr[0,4]
```

```python
# finding 30

arr[5,4]
```

```python
# slicing

slice = arr[0:2,1:3]

slice
```

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
slice = arr[3:,3:]

slice
```

```python
arr[:,2]
```

## **Boolean indexing**


**note** : ~ *while doing the boolean indexing we must keep in mind that both the array and the boolean index must be of the same shape*

```python
arr = np.arange(11,21)
arr
```

```python
bool_index = arr % 2 ==0
bool_index
```

```python
arr = arr[bool_index]
arr
```
