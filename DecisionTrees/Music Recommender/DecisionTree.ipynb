# Decision Tree: Music Recommender
Minimal: train a DecisionTree on music.csv and export a GraphViz .dot.

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
from joblib import dump

# Load data (ensure this file exists in the same folder)
music_data = pd.read_csv('music.csv')
X = music_data.drop(columns=['genre'])
y = music_data['genre']

model = DecisionTreeClassifier()
model.fit(X, y)

# Export tree and save model
tree.export_graphviz(model, out_file='music_recommender.dot', feature_names=X.columns.tolist(), class_names=[str(c) for c in sorted(y.unique())], label='all', rounded=True, filled=True)
dump(model, 'music_recommender.joblib')
print('Saved: music_recommender.joblib and music_recommender.dot')
```
